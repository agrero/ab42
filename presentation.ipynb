## Slide Code

### Introduction

```python
intro = SingleColumnSlide(
    title="Something Declarative About the Results ",
    content_block=TextContent(
        text="""
    <b>An Investigation into the Sequence Permuations of beta-Amyloid-42</b>
    By: Alex Guerrero
    """
    )
)
```

### Summary Stats Slides

```python
summary_stats_intro_slide = SingleColumnSlide(
    title="Row/Columnwise Summary Statistics",
    content_block=TextContent(
        """
        A collection of secondary element counts by both sequence and sequence order
        """
    )
)

#######

summary_columns_plot = pio.read_json(
    os.path.join("plots", "2dary_summary_cols.json")
)
summary_columns_slide = SingleColumnSlide(
    title="Raw Data",
    content_block=PlotlyContent(
        fig=summary_columns_plot
    )
)
summary_columns_normal_plot = pio.read_json(
    os.path.join("plots", "2dary_summary_cols_normal.json")
)
summary_columns_normal_slide = SingleColumnSlide(
    title="Robust Scaled",
    content_block=PlotlyContent(
        fig=summary_columns_normal_plot
    )
)
summary_rows_plot = pio.read_json(
    os.path.join("plots", "2dary_summary_rows.json")
)
summary_rows_slide = SingleColumnSlide(
    title="",
    content_block=PlotlyContent(
        fig=summary_rows_plot
    )
)

#######

summary_stats_takeaway_slide = SingleColumnSlide(
    title="Global Stats Takeaways",
    base_font_size="22px",
    content_block=TextContent("""
- The sequence space is mainly disordered

- Our initial assumptions of the tail sequences being always disordered is half correct as the 6 Amino acids making up the N terminus were more ordered than the rest of the sequence

- Interestingly we do see a differential ordering effect across the N and C termini by the C terminus being the most disordered region of the protein whereas the N termini is the most ordered.

- The C terminal residue is always disordered regardless of the sequence context it is in.
""")
)
```

### Motif Len 3 Slides

```python
m3_intro_slide = SingleColumnSlide(
    title="Motif Length 3 Sum Plots",
    content_block=TextContent("""

    <b>Say Something About the Motifs Here</b>
    # To Do
    I think we need to add some sort of introduction here on to how we got these numbers
    
    OR we just explain well with the first one

""")
)
m3_root = "motiflen_3"
m3_pdir = os.path.join("plots", m3_root)

m3_plots = [
    pio.read_json(os.path.join(m3_pdir, f"{m3_root}_{i}.json")) 
    for i in sec_elements
]

m3_slides = [
    SingleColumnSlide(title="",content_block=PlotlyContent(plot))
    for plot in m3_plots
]

m3_conclusion_slide = SingleColumnSlide(
    title="Motif Length 3 Takeaways",
    content_block=TextContent("""
    <b>Alpha Helix:</b> DAE and FRH (6 N terminal residues) show a significantly larger proclivity for helical character than the rest of the 3 amino acids pairings.

    <b>Coil:</b> Both termini show their proclivity for disordered configurations. This somewhat supports our initial hypothesis that the termini will be inherently disordered regardless of context.

    <b>Strand:</b> Interestingly here, in converse to our N termini the C terminus has the seemingly highest proclivity to adopt configurations. The comparison to the N terminus isn't complete as the Strand content has a significantly higher variance.  
    
    <b>310, Bridge:</b> I don't see how these fit in narratively due to their low counts and think they should be relegated to supplementary slides.

    <b>Turn:</b> I think this one doesn't fit in narratively similarly to the above, but the counts are significantly higher.
    """)
)
```
