```python
import plotly.io as pio

from slide_types.SingleColumnSlide import SingleColumnSlide
from slide_types.content.PlotlyContent import PlotlyContent
from slide_types.content.TextContent import TextContent

import os
```

```python
save = True
```

```python
intro = SingleColumnSlide(
    title="Something Declarative About the Results ",
    content_block=TextContent(
        text="""
    <b>An Investigation into the Sequence Permuations of beta-Amyloid-42</b>
    By: Alex Guerrero
    """
    )
)
```

```python
summary_stats_intro_slide = SingleColumnSlide(
    title="Row/Columnwise Summary Statistics",
    content_block=TextContent(
        """
        A collection of secondary element counts by both sequence and sequence order
        """
    )
)

#######

summary_columns_plot = pio.read_json(
    os.path.join("plots", "2dary_summary_cols.json")
)
summary_columns_slide = SingleColumnSlide(
    title="Raw Data",
    content_block=PlotlyContent(
        fig=summary_columns_plot
    )
)
summary_columns_normal_plot = pio.read_json(
    os.path.join("plots", "2dary_summary_cols_normal.json")
)
summary_columns_normal_slide = SingleColumnSlide(
    title="Robust Scaled",
    content_block=PlotlyContent(
        fig=summary_columns_normal_plot
    )
)
summary_rows_plot = pio.read_json(
    os.path.join("plots", "2dary_summary_rows.json")
)
summary_rows_slide = SingleColumnSlide(
    title="",
    content_block=PlotlyContent(
        fig=summary_rows_plot
    )
)

#######

summary_stats_takeaway_slide = SingleColumnSlide(
    title="Global Stats Takeaways",
    base_font_size="22px",
    content_block=TextContent("""
- The sequence space is mainly disordered

- Our initial assumptions of the tail sequences being always disordered is half correct as the 6 Amino acids making up the N terminus were more ordered than the rest of the sequence

- Interestingly we do see a differential ordering effect across the N and C termini by the C terminus being the most disordered region of the protein whereas the N termini is the most ordered.

- The C terminal residue is always disordered regardless of the sequence context it is in.
""")
)
```

```python
motif_sum_intro_slide = SingleColumnSlide(
    title="Motif Sum Plots V2",
    content_block=TextContent(
        text="<b>Each bar is colored by its Secondary Structure Type, and the ordering does not change</b>"
    )
)
motif_sum_3_plot = pio.read_json("plots/motiflen_3_v2.json")
motif_sum_3_slide = SingleColumnSlide(
    title="",
    content_block=PlotlyContent(
        fig = motif_sum_3_plot
    )
)
```
